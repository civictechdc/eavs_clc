```python
import polars as pl

data_2020 = pl.read_csv("../eavs_clc/data/cleaned/2020.csv", schema_overrides = {"fips_code": pl.String, "year": pl.String})
data_2022 = pl.read_csv("../eavs_clc/data/cleaned/2022.csv", schema_overrides = {"fips_code": pl.String, "year": pl.String})
```

```python
data_2020 = data_2020.with_columns(
    year = pl.lit(2020)
)

data_2022 = data_2022.with_columns(
    year = pl.lit(2022)
)
```

```python
combined = data_2020.vstack(data_2022)
```

```python
combined = combined.fill_null(strategy = "zero")
```

```python
combined_state = combined.group_by("state")
float_cols = [col for col, dtype in zip(combined.columns, combined.dtypes) 
              if dtype in [pl.Float32, pl.Float64, pl.Int32, pl.Int64]]
string_cols = [col for col, dtype in zip(combined.columns, combined.dtypes) 
              if dtype in [pl.String] and col != "state"]
combined_state = combined_state.agg(
    [pl.col(col).first() for col in string_cols] + 
    [pl.col(col).mean() for col in float_cols]
)

combined_county = combined.group_by("fips_code")
float_cols = [col for col, dtype in zip(combined.columns, combined.dtypes) 
              if dtype in [pl.Float32, pl.Float64, pl.Int32, pl.Int64]]
string_cols = [col for col, dtype in zip(combined.columns, combined.dtypes) 
              if dtype in [pl.String] and col != "fips_code"]
combined_county = combined_county.agg(
    [pl.col(col).first() for col in string_cols] + 
    [pl.col(col).mean() for col in float_cols]

)
```

```python
combined_state
```

```python
combined_county = combined_county.with_columns(
    (pl.when(pl.col("fips_code").str.len_chars() == 9)
    .then(pl.col("fips_code").str.zfill(10))
    .otherwise(pl.col("fips_code")))
    .alias("fips_code")
)
```

```python
combined_county
```

```python
column_list = [
    'total_registrations_received',
    'new_valid_registrations',
    'rejected_registrations',
    'total_forms_mail_fax_email',
    'new_registrations_mail_fax_email',
    'rejected_registrations_mail_fax_email',
    'total_forms_in_person',
    'new_registrations_in_person',
    'rejected_registrations_in_person',
    'total_forms_online',
    'new_registrations_online',
    'rejected_registrations_online',
    'total_forms_dmv',
    'new_registrations_dmv',
    'rejected_registrations_dmv',
    'total_forms_mandatory_nvra',
    'new_registrations_mandatory_nvra',
    'rejected_registrations_mandatory_nvra',
    'total_forms_disability_agency',
    'new_registrations_disability_agency',
    'rejected_registrations_disability_agency',
    'total_forms_armed_forces',
    'new_registrations_armed_forces',
    'rejected_registrations_armed_forces',
    'total_forms_discretionary_nvra',
    'new_registrations_discretionary_nvra',
    'rejected_registrations_discretionary_nvra',
    'total_forms_advocacy_groups',
    'new_registrations_advocacy_groups',
    'rejected_registrations_advocacy_groups',
    'year'              
]               

combined_state = combined_state.select('state', *column_list)
combined_county = combined_county.select("fips_code", "jurisdiction_name", *column_list)
```

```python
dataset_list = [combined_state, combined_county]
```

```python
for name, df in [("state", combined_state), ("county", combined_county)]:
    df = df.with_columns([
        (pl.col("rejected_registrations") / pl.col("total_registrations_received") * 100)
            .alias("percent_rejected_registrations_total"),
        (pl.col("rejected_registrations_mail_fax_email") / pl.col("new_registrations_mail_fax_email") * 100)
            .alias("percent_rejected_registrations_mail_fax_email"),
        (pl.col("rejected_registrations_in_person") / pl.col("new_registrations_in_person") * 100)
            .alias("percent_rejected_registrations_in_person"),
        (pl.col("rejected_registrations_online") / pl.col("new_registrations_online") * 100)
            .alias("percent_rejected_registrations_online"),
        (pl.col("rejected_registrations_dmv") / pl.col("new_registrations_dmv") * 100)
            .alias("percent_rejected_registrations_dmv"),
        (pl.col("rejected_registrations_mandatory_nvra") / pl.col("new_registrations_mandatory_nvra") * 100)
            .alias("percent_rejected_registrations_mandatory_nvra"),
        (pl.col("rejected_registrations_disability_agency") / pl.col("new_registrations_disability_agency") * 100)
            .alias("percent_rejected_registrations_disability_agency"),
        (pl.col("rejected_registrations_armed_forces") / pl.col("new_registrations_armed_forces") * 100)
            .alias("percent_rejected_registrations_armed_forces"),
        (pl.col("rejected_registrations_discretionary_nvra") / pl.col("new_registrations_discretionary_nvra") * 100)
            .alias("percent_rejected_registrations_discretionary_nvra"),
        (pl.col("rejected_registrations_advocacy_groups") / pl.col("new_registrations_advocacy_groups") * 100)
            .alias("percent_rejected_registrations_advocacy_groups"),
    ])

    if name == "state":
        combined_state = df
    else:
        combined_county = df
```

```python
combined_state = combined_state.with_columns(
    pl.col("percent_rejected_registrations_total")
      .rank(descending=True, method="ordinal")
      .alias("rejected_state_rank")
)

combined_county = combined_county.with_columns(
    pl.col("percent_rejected_registrations_total")
      .rank(descending=True, method="ordinal")
      .alias("rejected_state_rank")
)
```

```python
combined_state = combined_state.fill_nan(0)

combined_county = combined_county.fill_nan(0)
```

```python
combined_pie_chart = combined_state.melt(
    id_vars=["state"],
    value_vars=[
        "rejected_registrations_advocacy_groups",
        "rejected_registrations_armed_forces",
        "rejected_registrations_disability_agency",
        "rejected_registrations_discretionary_nvra",
        "rejected_registrations_dmv",
        "rejected_registrations_in_person",
        "rejected_registrations_mail_fax_email",
        "rejected_registrations_mandatory_nvra",
        "rejected_registrations_online",
    ],
    variable_name="rejection_type",
    value_name="count"
)

name_map = {
    "rejected_registrations_advocacy_groups": "Advocacy Groups",
    "rejected_registrations_armed_forces": "Armed Forces",
    "rejected_registrations_disability_agency": "Disability Agency",
    "rejected_registrations_discretionary_nvra": "Discretionary NVRA",
    "rejected_registrations_dmv": "DMV",
    "rejected_registrations_in_person": "In Person",
    "rejected_registrations_mail_fax_email": "Mail/Fax/Email",
    "rejected_registrations_mandatory_nvra": "Mandatory NVRA",
    "rejected_registrations_online": "Online",
}

combined_pie_chart = combined_pie_chart.with_columns(
    pl.col("rejection_type").map_elements(lambda x: name_map.get(x, x)).alias("rejection_type")
)
```

```python
combined_pie_chart_county = combined_county.melt(
    id_vars=["fips_code"],
    value_vars=[
        "rejected_registrations_advocacy_groups",
        "rejected_registrations_armed_forces",
        "rejected_registrations_disability_agency",
        "rejected_registrations_discretionary_nvra",
        "rejected_registrations_dmv",
        "rejected_registrations_in_person",
        "rejected_registrations_mail_fax_email",
        "rejected_registrations_mandatory_nvra",
        "rejected_registrations_online",
    ],
    variable_name="rejection_type",
    value_name="count"
)

name_map = {
    "rejected_registrations_advocacy_groups": "Advocacy Groups",
    "rejected_registrations_armed_forces": "Armed Forces",
    "rejected_registrations_disability_agency": "Disability Agency",
    "rejected_registrations_discretionary_nvra": "Discretionary NVRA",
    "rejected_registrations_dmv": "DMV",
    "rejected_registrations_in_person": "In Person",
    "rejected_registrations_mail_fax_email": "Mail/Fax/Email",
    "rejected_registrations_mandatory_nvra": "Mandatory NVRA",
    "rejected_registrations_online": "Online",
}

combined_pie_chart_county = combined_pie_chart_county.with_columns(
    pl.col("rejection_type").map_elements(lambda x: name_map.get(x, x)).alias("rejection_type")
)
```

```python
print(f"Total County Fields: {combined_pie_chart_county.select(pl.len()).item():,}")
print(f"Non-Null County Fields: {combined_pie_chart_county.filter(pl.col("count") != 0).select(pl.len()).item():,}")
```
